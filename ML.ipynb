```python

nums = list(map(int, input("Enter numbers: ").split()))


if len(nums) % 2 != 0:
    print("odd")
else:

    for i in range(0, len(nums), 2):
        nums[i], nums[i+1] = nums[i+1], nums[i]

    print(" ".join(map(str, nums)))
```
