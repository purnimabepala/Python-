```python
n, m = map(int, input().split())
edges = []
for _ in range(m):
    u, v, w = map(int, input().split())
    edges.append((w, u, v))
edges.sort()
parent = list(range(n))

def find(x):
    while parent[x] != x:
        x = parent[x]
    return x

def union(x, y):
    parent[find(x)] = find(y)
mst_weight = 0
edges_used = 0

for w, u, v in edges:
    if find(u) != find(v):
        union(u, v)
        mst_weight += w
        edges_used += 1
    if edges_used == n-1:
        break
print("MST weight =", mst_weight)
```
